```python
# Create a 1D NumPy array
brightness = np.array(df["Normalized_Brightness"])

print("1D Array:")
print(brightness[:10])

# Create a 2D NumPy array
numerical_data = np.array(
    df[
        [
            "Normalized_Brightness",
            "Transit_Depth",
            "Orbital_Period_Days",
            "Transit_Duration_Hours"
        ]
    ]
)

print("\n2D Array:")
print(numerical_data[:5])

print("\nShape of 1D array:", brightness.shape)
print("Shape of 2D array:", numerical_data.shape)
```

```python
# Take first 10 brightness values
a = brightness[:10]

# Create another array
b = np.ones(10)

print("Array A:")
print(a)

print("\nArray B:")
print(b)

print("\nAddition:")
print(a + b)

print("\nSubtraction:")
print(a - b)

print("\nMultiplication:")
print(a * b)

print("\nDivision:")
print(a / b)

print("\nTotal:", np.sum(a))
print("Average:", np.mean(a))
print("Minimum:", np.min(a))
print("Maximum:", np.max(a))
```

```python
numeric_columns = [
    "Normalized_Brightness",
    "Transit_Depth",
    "Orbital_Period_Days",
    "Transit_Duration_Hours"
]

for column in numeric_columns:
    print("\n", column)
    print("Total:", df[column].sum())
    print("Average:", df[column].mean())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
```

```python
brightness_np = np.array(df["Normalized_Brightness"])

numpy_mean = np.mean(brightness_np)
pandas_mean = df["Normalized_Brightness"].mean()

numpy_total = np.sum(brightness_np)
pandas_total = df["Normalized_Brightness"].sum()

print("NumPy Mean:", numpy_mean)
print("Pandas Mean:", pandas_mean)

print("\nNumPy Total:", numpy_total)
print("Pandas Total:", pandas_total)

print("\nMean values equal:", np.isclose(numpy_mean, pandas_mean))
print("Total values equal:", np.isclose(numpy_total, pandas_total))
```

```python
df["Observation_Date"] = pd.to_datetime(df["Observation_Date"])

print(df["Observation_Date"].dtype)
print(df.head())
```

```python
daily_summary = df.groupby("Observation_Date").agg(
    Observation_Count=("Planet", "count"),
    Average_Brightness=("Normalized_Brightness", "mean"),
    Average_Transit_Depth=("Transit_Depth", "mean")
)

daily_summary
```
